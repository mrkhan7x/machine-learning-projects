# Big Mart Sales Prediction using Machine Learning

In this notebook, we build a Machine Learning regression model to predict **Item Outlet Sales** across retail stores using **Label Encoding** and **Random Forest Regressor**.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn import metrics
```

### Data Collection and Processing


```python
# Loading the dataset
big_mart_data = pd.read_csv('Train.csv')
big_mart_data.head()
```

```python
# Shape of dataset
print("Rows and Columns:", big_mart_data.shape)
big_mart_data.info()
```

```python
# Checking missing values
big_mart_data.isnull().sum()
```

### Handling Missing Values
* **Mean** for `Item_Weight`
* **Mode** for `Outlet_Size`


```python
# Mean imputation for Item_Weight
big_mart_data['Item_Weight'].fillna(big_mart_data['Item_Weight'].mean(), inplace=True)

# Mode imputation for Outlet_Size
mode_outlet_size = big_mart_data['Outlet_Size'].mode()[0]
big_mart_data['Outlet_Size'].fillna(mode_outlet_size, inplace=True)

# Check missing values again
big_mart_data.isnull().sum()
```

### Data Visualization


```python
plt.figure(figsize=(6, 4))
sns.histplot(big_mart_data['Item_Weight'], kde=True, color='purple')
plt.title('Item Weight Distribution')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(big_mart_data['Item_MRP'], kde=True, color='teal')
plt.title('Item MRP Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(x='Outlet_Establishment_Year', data=big_mart_data, palette='Set2')
plt.title('Outlet Establishment Years')
plt.show()
```

### Categorical Encoding with LabelEncoder


```python
encoder = LabelEncoder()
cat_cols = ['Item_Fat_Content', 'Item_Type', 'Outlet_Size', 'Outlet_Location_Type', 'Outlet_Type']
for col in cat_cols:
    big_mart_data[col] = encoder.fit_transform(big_mart_data[col].astype(str))

big_mart_data.head()
```

### Splitting Features and Target


```python
X = big_mart_data.drop(columns=['Item_Identifier', 'Outlet_Identifier', 'Item_Outlet_Sales'], axis=1)
Y = big_mart_data['Item_Outlet_Sales']
print(X.shape, Y.shape)
```

### Train-Test Split


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
print("Train:", X_train.shape, "Test:", X_test.shape)
```

### Model Training --> Random Forest Regressor


```python
model = RandomForestRegressor(n_estimators=100, random_state=2)
model.fit(X_train, Y_train)
```

### Model Evaluation


```python
test_pred = model.predict(X_test)
r2 = metrics.r2_score(Y_test, test_pred)
mae = metrics.mean_absolute_error(Y_test, test_pred)
print('R Squared Score : ', round(r2, 4))
print('Mean Absolute Error : ', round(mae, 2))
```

### Building a Predictive System


```python
sample_features = X_test.iloc[0].values.reshape(1, -1)
predicted_sales = model.predict(sample_features)
print('=' * 45)
print(f"  PREDICTED ITEM OUTLET SALES: ${predicted_sales[0]:.2f}")
print('=' * 45)
```
