# Information Retrieval & Semantic Search Engine

In this notebook, we build an **Information Retrieval & Vector Search Engine** using **TF-IDF Vectorization** and **Cosine Similarity**. We index a technical corpus and compute ranked search results for natural language queries.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

### Knowledge Base (Corpus Collection)


```python
corpus = [
    "Machine learning algorithms identify patterns in historical data to make predictive forecasts.",
    "Deep learning and artificial neural networks are inspired by biological brain architectures.",
    "Supervised learning models require labeled datasets containing input features and target outcomes.",
    "Natural language processing allows computer software to understand and generate human text.",
    "Gradient descent optimizes model weights by minimizing the loss function iteratively.",
    "Convolutional neural networks excel at computer vision tasks like image classification and detection.",
    "FastAPI and Flask are popular web frameworks for building backend APIs in Python.",
    "Relational SQL databases use structured schemas and keys to ensure transactional integrity.",
    "Docker containers package application code and dependencies for reliable deployment across servers.",
    "Ensemble methods like Random Forests and XGBoost combine multiple models to boost accuracy."
]

df_corpus = pd.DataFrame({"Doc_ID": range(1, len(corpus) + 1), "Document": corpus})
df_corpus
```

### Vectorizing Corpus with TF-IDF


```python
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(corpus)

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Total Vocabulary Size:", len(vectorizer.get_feature_names_out()))
```

### Document Similarity Heatmap


```python
sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(sim_matrix, annot=True, fmt='.2f', cmap='YlGnBu', xticklabels=range(1, 11), yticklabels=range(1, 11))
plt.title('Document-to-Document Cosine Similarity Matrix')
plt.xlabel('Document ID')
plt.ylabel('Document ID')
plt.show()
```

### Building the Semantic Search Function


```python
def search_corpus(query, top_k=3):
    # 1. Vectorize query
    query_vec = vectorizer.transform([query])
    
    # 2. Compute similarity against all documents
    scores = cosine_similarity(query_vec, tfidf_matrix)[0]
    
    # 3. Sort indices descending
    ranked_indices = np.argsort(scores)[::-1][:top_k]
    
    print('=' * 60)
    print(f"  SEARCH QUERY: '{query}'")
    print('=' * 60)
    for rank, idx in enumerate(ranked_indices, 1):
        score = scores[idx]
        print(f"  {rank}. [Score: {round(score * 100, 1)}%] Doc #{idx + 1}: {corpus[idx]}")
    print('=' * 60)

# Test Query 1
search_corpus("neural networks and deep learning architectures", top_k=3)

# Test Query 2
search_corpus("how to deploy python web APIs", top_k=3)
```
