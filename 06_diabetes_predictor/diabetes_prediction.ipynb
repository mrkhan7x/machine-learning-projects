# Diabetes Prediction using Logistic Regression

In this notebook, we build an end-to-end Machine Learning model to predict whether a patient has diabetes based on diagnostic measurements using **Logistic Regression**.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### Data Collection and Analysis


```python
# Loading the diabetes dataset to a pandas DataFrame
diabetes_data = pd.read_csv('diabetes.csv')
diabetes_data.head()
```

```python
# Number of rows and columns
diabetes_data.shape
```

```python
# Statistical measures of the data
diabetes_data.describe()
```

```python
# Checking missing values
diabetes_data.isnull().sum()
```

```python
# Checking distribution of target variable
diabetes_data['Outcome'].value_counts()
```

### Label Meaning:
* **0 --> Non-Diabetic**
* **1 --> Diabetic**


```python
diabetes_data.groupby('Outcome').mean()
```

### Data Visualization


```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=diabetes_data, palette=['#3498db', '#e74c3c'])
plt.title('Outcome Distribution: Non-Diabetic (0) vs Diabetic (1)')
plt.xticks(ticks=[0, 1], labels=['Non-Diabetic (0)', 'Diabetic (1)'])
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(diabetes_data.corr(), annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Diabetes Features')
plt.show()
```

### Separating Features and Target


```python
X = diabetes_data.drop(columns='Outcome', axis=1)
Y = diabetes_data['Outcome']
print(X.shape, Y.shape)
```

### Splitting the Data into Training Data & Test Data


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
print("Total:", X.shape, "Train:", X_train.shape, "Test:", X_test.shape)
```

### Model Training --> Logistic Regression


```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

### Model Evaluation


```python
# Accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy on training data : ', round(training_data_accuracy * 100, 2), '%')

# Accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy on test data     : ', round(test_data_accuracy * 100, 2), '%')
```

```python
# Confusion Matrix Heatmap
cm = confusion_matrix(Y_test, X_test_prediction)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Non-Diabetic', 'Diabetic'], yticklabels=['Non-Diabetic', 'Diabetic'])
plt.title('Confusion Matrix on Test Set')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### Building a Predictive System


```python
input_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)

# Changing input_data to numpy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshape array for predicting one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

prediction = model.predict(input_data_reshaped)
print("Model Prediction:", prediction)

if prediction[0] == 0:
    print('>>> Result: The patient is Non-Diabetic.')
else:
    print('>>> Result: The patient has Diabetes.')
```
