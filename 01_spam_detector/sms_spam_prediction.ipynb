# SMS Spam Detection using Machine Learning

In this notebook, we build an end-to-end Machine Learning model to classify SMS messages as either **Ham (Legitimate)** or **Spam** using **TF-IDF Vectorization** and **Logistic Regression**.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### Data Collection & Preprocessing


```python
# Sample SMS corpus representing real-world distribution
data = {
    'Category': ['ham', 'spam', 'ham', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham',
                 'ham', 'spam', 'ham', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham'],
    'Message': [
        'Hey bro, are we still meeting for lunch today at the cafeteria?',
        'WINNER!! As a valued customer you have been selected to receive £900 prize reward! Call now to claim!',
        'Can you please send me the notes from today machine learning lecture?',
        'I will be reaching home in 20 minutes, please make sure the door is unlocked.',
        'URGENT! Your mobile number was awarded a $1000 gift card. Reply YES to confirm delivery.',
        'Sounds good to me, see you tomorrow morning at 9am.',
        'Free entry in a weekly competition to win cash prizes! Text WIN to 80085 to enter.',
        'Hey, did you finish the assignment for semester 5 statistics?',
        'Congratulations! You have won a guaranteed cash bonus. Click the link to claim immediately.',
        'Please call mom when you get a chance, she was asking about your schedule.',
        'Are you coming to the football match tonight or staying at the dorm?',
        'PRIVATE! Your account has a pending payment bonus. Call 09061701461 to claim your reward.',
        'Let me know when you are free for a quick call regarding the project.',
        'The weather is so nice today, let us go for a walk in the evening.',
        'You have won 1000000 dollars in our international lucky draw! Reply with bank details to transfer.',
        'Thanks for helping me with the linear regression code yesterday.',
        'Claim your free holiday package to Maldives now! Call 08000930705 today only.',
        'I will submit the assignment on GitHub tonight after testing the models.',
        'Exclusive loan offer! Get an instant personal loan approved in 5 minutes with zero interest.',
        'Don\'t forget to bring your laptop charger tomorrow for the lab session.'
    ]
}

mail_data = pd.DataFrame(data)
mail_data.head()
```

```python
# Checking rows and columns
print("Shape of dataset:", mail_data.shape)
```

```python
# Checking missing values
mail_data.isnull().sum()
```

### Label Encoding

* **Spam --> 0**
* **Ham --> 1**


```python
mail_data.loc[mail_data['Category'] == 'spam', 'Category'] = 0
mail_data.loc[mail_data['Category'] == 'ham', 'Category'] = 1
mail_data['Category'] = mail_data['Category'].astype(int)
mail_data.head()
```

```python
# Checking class distribution
mail_data['Category'].value_counts()
```

### Data Visualization


```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Category', data=mail_data, palette=['#e74c3c', '#2ecc71'])
plt.title('Distribution of Ham vs Spam Messages')
plt.xticks(ticks=[0, 1], labels=['Spam (0)', 'Ham (1)'])
plt.xlabel('Message Type')
plt.ylabel('Count')
plt.show()
```

### Separating Features and Target


```python
X = mail_data['Message']
Y = mail_data['Category']
print(X.shape, Y.shape)
```

### Splitting Data into Training and Test Data


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=3)
print("Total:", X.shape, "Train:", X_train.shape, "Test:", X_test.shape)
```

### Feature Extraction (TF-IDF Vectorization)


```python
# Transform text data to feature vectors
feature_extraction = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)

X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)

print("Feature matrix shape:", X_train_features.shape)
```

### Model Training --> Logistic Regression


```python
model = LogisticRegression()
model.fit(X_train_features, Y_train)
```

### Model Evaluation


```python
# Accuracy on training data
X_train_prediction = model.predict(X_train_features)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy on training data : ', round(training_data_accuracy * 100, 2), '%')

# Accuracy on test data
X_test_prediction = model.predict(X_test_features)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy on test data : ', round(test_data_accuracy * 100, 2), '%')
```

### Building a Predictive System


```python
input_mail = ["WINNER!! You have won a guaranteed cash prize reward of £1000! Call immediately to claim your money!"]

# Convert text to feature vectors
input_data_features = feature_extraction.transform(input_mail)

# Making prediction
prediction = model.predict(input_data_features)
print("Prediction output:", prediction)

if prediction[0] == 1:
    print('>>> Result: HAM MAIL (Legitimate Message)')
else:
    print('>>> Result: SPAM MAIL (Junk / Fraudulent Message)')
```

```python
# Test with a normal ham message
input_mail_2 = ["Hey Roman, are we studying at the library tonight for the midterm exam?"]
input_features_2 = feature_extraction.transform(input_mail_2)
pred_2 = model.predict(input_features_2)

if pred_2[0] == 1:
    print('>>> Result: HAM MAIL (Legitimate Message)')
else:
    print('>>> Result: SPAM MAIL (Junk / Fraudulent Message)')
```
