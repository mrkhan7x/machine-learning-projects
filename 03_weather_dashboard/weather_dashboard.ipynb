# Weather Data Processing & Visualization Dashboard

In this notebook, we build a **Weather Analytics Dashboard** in Python. We process weather metrics across global cities, analyze correlations between atmospheric variables, and visualize temperature, humidity, and pressure.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Weather Data Collection & Structuring


```python
weather_data = {
    'City': ['London', 'Tokyo', 'New York', 'Paris', 'Karachi', 'Dubai', 'Singapore', 'Sydney', 'Cairo', 'Toronto'],
    'Country': ['UK', 'Japan', 'USA', 'France', 'Pakistan', 'UAE', 'Singapore', 'Australia', 'Egypt', 'Canada'],
    'Temperature_C': [14.5, 18.2, 22.0, 16.8, 33.5, 38.0, 31.2, 19.5, 29.0, 11.2],
    'Humidity_Pct': [78, 65, 55, 72, 60, 45, 85, 62, 38, 68],
    'Pressure_hPa': [1015, 1012, 1018, 1014, 1008, 1006, 1010, 1020, 1011, 1022],
    'Wind_Speed_mps': [4.2, 3.1, 5.5, 3.8, 6.2, 4.8, 2.5, 5.0, 4.0, 6.8],
    'Condition': ['Cloudy', 'Clear', 'Sunny', 'Rain', 'Sunny', 'Sunny', 'Rain', 'Clear', 'Sunny', 'Windy']
}

df_weather = pd.DataFrame(weather_data)
df_weather.head(10)
```

```python
# Statistical Summary
df_weather.describe()
```

### Data Visualization

#### 1. Temperature Comparison across Global Cities


```python
plt.figure(figsize=(10, 5))
sns.barplot(x='City', y='Temperature_C', data=df_weather, palette='coolwarm')
plt.title('Current Temperature (°C) Across Cities', fontsize=14)
plt.xlabel('City')
plt.ylabel('Temperature (°C)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#### 2. Temperature vs. Humidity Relationship


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Temperature_C', y='Humidity_Pct', hue='Condition', data=df_weather, s=150, palette='Set2')
plt.title('Temperature vs. Humidity by Weather Condition', fontsize=14)
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.show()
```

#### 3. Correlation Heatmap


```python
numeric_cols = df_weather.select_dtypes(include=[np.number])
plt.figure(figsize=(7, 5))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Atmospheric Metrics Correlation Heatmap')
plt.show()
```

### Interactive City Weather Lookup System


```python
def get_weather_card(city_name):
    match = df_weather[df_weather['City'].str.lower() == city_name.lower()]
    if len(match) == 0:
        print(f"[NOT FOUND] No data recorded for '{city_name}'.")
        return
    row = match.iloc[0]
    print('=' * 45)
    print(f"  WEATHER REPORT: {row['City'].upper()}, {row['Country']}")
    print('=' * 45)
    print(f"  * Condition:       {row['Condition']}")
    print(f"  * Temperature:     {row['Temperature_C']} °C ({round(row['Temperature_C']*9/5 + 32, 1)} °F)")
    print(f"  * Humidity:        {row['Humidity_Pct']} %")
    print(f"  * Pressure:        {row['Pressure_hPa']} hPa")
    print(f"  * Wind Speed:      {row['Wind_Speed_mps']} m/s")
    print('=' * 45)

get_weather_card('Karachi')
get_weather_card('Tokyo')
```
