# Gold Price Prediction using Random Forest Regressor

In this notebook, we build an accurate Machine Learning regression model to predict **Gold Prices (GLD ETF)** based on financial market indices using **Random Forest Regressor**.


### Importing the Dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn import metrics
```

### Data Collection and Processing


```python
# Loading the csv data to a Pandas DataFrame
gold_data = pd.read_csv('gld_price_data.csv')
gold_data.head()
```

```python
# Print last 5 rows of the dataframe
gold_data.tail()
```

```python
# Number of rows and columns
gold_data.shape
```

```python
# Getting some basic informations about the data
gold_data.info()
```

```python
# Checking the number of missing values
gold_data.isnull().sum()
```

```python
# Getting the statistical measures of the data
gold_data.describe()
```

### Correlation Analysis

* **Positive Correlation**
* **Negative Correlation**


```python
correlation = gold_data.drop('Date', axis=1).corr()

# Constructing a heatmap to understand the correlation
plt.figure(figsize=(8, 8))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f', annot=True, annot_kws={'size': 8}, cmap='Blues')
plt.title('Correlation Heatmap of Financial Indices')
plt.show()
```

```python
# Correlation values of GLD with other assets
print(correlation['GLD'])
```

```python
# Checking the distribution of the GLD Price
plt.figure(figsize=(7, 4))
sns.histplot(gold_data['GLD'], color='green', kde=True)
plt.title('Distribution of Gold (GLD) Price')
plt.show()
```

### Splitting the Features and Target


```python
X = gold_data.drop(['Date', 'GLD'], axis=1)
Y = gold_data['GLD']
print(X)
```

### Splitting into Training Data and Test Data


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
print("Total:", X.shape, "Train:", X_train.shape, "Test:", X_test.shape)
```

### Model Training --> Random Forest Regressor


```python
regressor = RandomForestRegressor(n_estimators=100)

# Training the model
regressor.fit(X_train, Y_train)
```

### Model Evaluation


```python
# Prediction on Test Data
test_data_prediction = regressor.predict(X_test)

# R squared error
error_score = metrics.r2_score(Y_test, test_data_prediction)
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
print("R squared error : ", round(error_score, 4))
print("Mean Absolute Error : ", round(mae, 4))
```

### Compare the Actual Values and Predicted Values in a Plot


```python
Y_test_list = list(Y_test)

plt.figure(figsize=(10, 5))
plt.plot(Y_test_list, color='blue', label='Actual Value', alpha=0.7)
plt.plot(test_data_prediction, color='green', label='Predicted Value', alpha=0.7)
plt.title('Actual Price vs Predicted Price')
plt.xlabel('Number of values')
plt.ylabel('GLD Price')
plt.legend()
plt.show()
```

### Building a Predictive System


```python
input_data = (1447.160034, 78.470001, 15.1800, 1.471692)

# Changing input_data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshaping the array
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

prediction = regressor.predict(input_data_reshaped)
print('=' * 45)
print(f"  PREDICTED GLD PRICE: ${prediction[0]:.2f}")
print('=' * 45)
```
