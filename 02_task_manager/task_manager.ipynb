# CLI Task Manager Application

In this notebook, we build an interactive **Task Manager** in Python using **Object-Oriented Programming (OOP)** and **JSON persistence**. We demonstrate creating, completing, filtering, and saving tasks to disk.


### Importing the Dependencies


```python
import os
import json
import pandas as pd
from datetime import datetime
```

### Designing the Task & TaskManager Classes


```python
class Task:
    def __init__(self, task_id, title, category="General", completed=False):
        self.id = task_id
        self.title = title
        self.category = category
        self.completed = completed
        self.created_at = datetime.now().strftime("%Y-%m-%d %H:%M")
        
    def to_dict(self):
        return {
            "id": self.id,
            "title": self.title,
            "category": self.category,
            "completed": self.completed,
            "created_at": self.created_at
        }

class TaskManager:
    def __init__(self, filename="tasks.json"):
        self.filename = filename
        self.tasks = []
        self.load_tasks()
        
    def add_task(self, title, category="General"):
        task_id = len(self.tasks) + 1
        task = Task(task_id, title, category)
        self.tasks.append(task)
        print(f"[ADDED] Task #{task_id}: '{title}' ({category})")
        self.save_tasks()
        return task
        
    def complete_task(self, task_id):
        for task in self.tasks:
            if task.id == task_id:
                task.completed = True
                print(f"[COMPLETED] Task #{task_id}: '{task.title}' marked as done!")
                self.save_tasks()
                return True
        print(f"[ERROR] Task #{task_id} not found.")
        return False
        
    def get_dataframe(self):
        return pd.DataFrame([t.to_dict() for t in self.tasks])
        
    def save_tasks(self):
        data = [t.to_dict() for t in self.tasks]
        with open(self.filename, "w") as f:
            json.dump(data, f, indent=4)
            
    def load_tasks(self):
        if os.path.exists(self.filename):
            with open(self.filename, "r") as f:
                data = json.load(f)
                self.tasks = [Task(d["id"], d["title"], d.get("category", "General"), d["completed"]) for d in data]
```

### Initializing and Adding Tasks


```python
manager = TaskManager("tasks.json")
manager.add_task("Study Probability & Statistics Chapter 6", "Academics")
manager.add_task("Train 5-Model Ensemble for GCI World 2026", "AI Competition")
manager.add_task("Refactor Machine Learning Portfolio GitHub Repo", "Portfolio")
manager.add_task("Automate client outreach workflow in n8n", "Agency")
```

### Visualizing Task Status Table


```python
df_tasks = manager.get_dataframe()
df_tasks['Status'] = df_tasks['completed'].apply(lambda x: '✅ Done' if x else '⏳ Pending')
df_tasks[['id', 'title', 'category', 'Status', 'created_at']]
```

### Completing and Managing Tasks


```python
# Mark task 1 and task 3 as completed
manager.complete_task(1)
manager.complete_task(3)

# Updated view
df_updated = manager.get_dataframe()
df_updated['Status'] = df_updated['completed'].apply(lambda x: '✅ Done' if x else '⏳ Pending')
df_updated[['id', 'title', 'category', 'Status']]
```
